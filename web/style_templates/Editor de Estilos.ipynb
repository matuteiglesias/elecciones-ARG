### Basename

````python
### FILL
def replace_fill_color_basename(style_object, new_basename):
    fill_color = style_object["paint"]["fill-color"]
    if (
        isinstance(fill_color, list)
        and len(fill_color) >= 3
        and fill_color[0] == "interpolate"
        and fill_color[1] == ["linear"]
        and isinstance(fill_color[2], list)
        and len(fill_color[2]) >= 2
        and fill_color[2][0] == "get"
    ):
        fill_color[2][1] = new_basename
        style_object["paint"]["fill-color"] = fill_color

    return style_object

### TEXT
def replace_text_field_basename(style_object, new_basename):
    style_object["layout"]["text-field"] = [
        "concat",
        ["to-string", ["/", ["round", ["*", 1000, ["get", new_basename]]], 10]],
        "%"
    ]
    return style_object
````

### Zoom Opacity

````python
### FILL
def replace_fill_opacity_zoom(style_object, new_zoom_values):


    fill_opacity = style_object["paint"]["fill-opacity"]
    if (
        isinstance(fill_opacity, list)
        and fill_opacity[0] == "interpolate"
        and fill_opacity[1] == ["linear"]
        and fill_opacity[2] == ["zoom"]
    ):
        fill_opacity = fill_opacity[:3] + new_zoom_values
        style_object["paint"]["fill-opacity"] = fill_opacity

    return style_object

# updated_style_object = replace_fill_opacity_zoom(edited_object, [1, 2, 3])

### TEXT
def replace_text_opacity_zoom(style_object, new_zoom_values):
    text_opacity = style_object["paint"]["text-opacity"]
    if (
        isinstance(text_opacity, list)
        and text_opacity[0] == "interpolate"
        and text_opacity[1] == ["linear"]
        and text_opacity[2] == ["zoom"]
    ):
        text_opacity = text_opacity[:3] + new_zoom_values
        style_object["paint"]["text-opacity"] = text_opacity

    return style_object
````

````python
import colorsys

# Hex
# def generate_color_palette(base_color, num_steps):
#     # Convert the base color from hex to RGB
#     r, g, b = tuple(int(base_color[i:i+2], 16) for i in (0, 2, 4))
    
#     # Generate the RGB colors for the palette in reverse order
#     colors = []
#     for i in range(num_steps - 1, -1, -1):
#         percent = i / (num_steps - 1)
#         rgb = colorsys.rgb_to_hls(r/255, g/255, b/255)
#         new_rgb = colorsys.hls_to_rgb(rgb[0], percent, rgb[2])
#         new_color = tuple(int(c * 255) for c in new_rgb)
#         colors.append(new_color)
    
#     # Convert the RGB colors to hex in reverse order
#     palette = {}
#     for i, rgb in enumerate(colors[::-1]):
#         hex_code = '#%02x%02x%02x' % rgb
#         percent = i * (100 / (num_steps - 1))
#         palette[percent] = hex_code
    
#     # Add the endpoints at 0% and 100%
#     palette[0] = '#ffffff'
#     palette[100] = '#000000'
    
#     return palette

# # RGB
# def generate_color_palette(base_color, num_steps):
#     r, g, b = base_color

#     palette = {}
#     for i in range(num_steps):
#         percent = i / (num_steps - 1)
#         rgba = (r, g, b, percent)
#         palette[percent] = rgba

#     return palette

# def generate_color_palette(base_color, num_steps, vmin=0, vmax=1):
#     r, g, b = base_color

#     palette = {}
#     for i in range(num_steps):
#         percent = vmin + (i / (num_steps - 1)) * (vmax - vmin)
#         rgba = (r, g, b, percent)
#         palette[percent] = rgba

#     return palette



def replace_fill_color_values(style_object, base_color, vmin=0, vmax=1):
    r, g, b = base_color

    color_palette = {v: (r, g, b, [0, 1][i]) for i, v in enumerate([vmin, vmax])}

    fill_color = style_object["paint"]["fill-color"]
    if (
        isinstance(fill_color, list)
        and fill_color[0] == "interpolate"
        and fill_color[1] == ["linear"]
        and isinstance(fill_color[2], list)
        and fill_color[2][0] == "get"
    ):
        new_fill_color = ["interpolate", ["linear"], fill_color[2]]
        for v in [vmin, vmax]:
            rgba = color_palette[v]
            new_fill_color.extend([v, f"rgba{rgba}"])

        style_object["paint"]["fill-color"] = new_fill_color

    return style_object

````
