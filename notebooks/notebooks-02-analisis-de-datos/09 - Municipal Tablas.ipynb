En esta eleccion, puede que esta situacion no se repita de la misma forma ya que el candidato es Massa y se supone que el nucleo duro kirchnerista va por lo menos a aportar el piso de los 20000 votos.

```python
df = pd.read_csv('./municipal/historico_listas_zarate.csv')[['año', 'eleccion_tag', 'cargo_nombre', 'agrupacion_nombre', 'lista_nombre', 'lista_votos', 'cluster']]
df['lista_votos'] = df['lista_votos'].round(-2)
titles = ['Sector FpV', 'Sector Cambiemos', 'Sector Massa']

for cluster in [1, 2, 3]:
    # Display markdown title of the table
    display(Markdown(f'#### {titles[cluster - 1]}'))

    show = df.loc[df.cluster == cluster]
    show = show.drop('cluster', axis = 1)
    show = show.loc[show.cargo_nombre.isin(['Diputado Nacional', 'Presidente'])]
    show = show.sort_values(['año', 'cargo_nombre'])
    show = show.reset_index(drop = True)
    display(show)
```

#### Circuitos Electorales Zarate

```python
import geopandas as gpd
# filter UserWarning
import warnings
warnings.filterwarnings('ignore')

circuitos_gdf = gpd.read_file('./../datos/geojson/circs_ref_simplified_1.geojson')

zarate = circuitos_gdf.loc[circuitos_gdf.NAM == 'Zárate']
zarate = zarate[['circuito_id', 'circuito', 'geometry']]

tamano_circs = pd.read_csv('./municipal/tamano_circs.csv')
zarate = zarate.merge(tamano_circs.reset_index())
```

**Mapa de los circuitos electorales y cuanto pesan dentro del partido**

Lima tiene solo un 9% de los votantes, Zarate el 90%, las zonas rurales no influyen en el resultado

```python
import matplotlib.pyplot as plt
ax = zarate.plot('Porcentaje', vmin=0, vmax=30, cmap='Greys', edgecolor='black', figsize = (8, 8))

# Add annotations in black
for x, y, label in zip(zarate.geometry.centroid.x, zarate.geometry.centroid.y, zarate['Porcentaje']):
    ax.text(x, y, f'{int(label)}%', color='black', ha='center', va='center')

offset_x = .001; offset_y = .001
# Add annotations in white (slightly offset for visibility)
for x, y, label in zip(zarate.geometry.centroid.x, zarate.geometry.centroid.y, zarate['Porcentaje']):
    ax.text(x + offset_x, y + offset_y, f'{int(label)}%', color='white', ha='center', va='center')

# Show the plot
plt.axis('off')
plt.show()
```
